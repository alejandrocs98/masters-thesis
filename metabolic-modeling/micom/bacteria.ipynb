```python
agora103 = pd.read_table('agora103_new_taxa.tsv', sep='\t', index_col=0)
agora103.drop('file', axis=1, inplace=True)
agora103.index.name = 'name'
agora103
```

```python
lf0 = pd.read_table('lf0_data.tsv')
lf0
```

```python
bacteria = lf0[lf0['sample_id'] == 'mouse_01'].copy()
bacteria.drop(['sample_id', 'abundance', 'genus', 'species'], axis=1, inplace=True)
bacteria = bacteria.loc[:, ['id', 'file']]
bacteria['name'] = bacteria['file'].apply(lambda x: x.split('/')[-1]).replace('.xml', '', regex=True)
bacteria['name'] = bacteria['name'].apply(lambda x: x.replace('Bacteroides_cellulosilyticus_WH2', 'Bacteroides_cellulosilyticus_DSM_14838'))
bacteria.set_index('name', inplace=True)
# bacteria.to_csv('bacteria_models.tsv', sep='\t', index=False)
bacteria
```

```python
merged = pd.merge(bacteria, agora103, left_index=True, right_index=True)
merged.reset_index(inplace=True)
merged.drop(['name', 
             'organism', 
             'seedid', 
             'draftcreator', 
             'draftcreated', 
             'ncbiid', 
             'platform', 
             'kbaseid',
             'genes', 
             'reactions', 
             'metabolites', 
             ], axis=1, inplace=True)
merged['species'] = merged['species'].apply(lambda x: x.replace('[', '').replace(']', ''))
merged['strain'] = merged['strain'].apply(lambda x: x.replace('Bacteroides cellulosilyticus DSM 14838', 'Bacteroides cellulosilyticus WH2'))
merged.set_index('id', inplace=True)
merged = merged.loc[:, ['kingdom', 'phylum', 'class', 'order', 'family', 'genus', 'species', 'strain', 'oxygenstat', 'metabolism', 'gram', 'mtype', 'file']]
merged.to_csv('bacteria_models.tsv', sep='\t', index=True)
merged
```
