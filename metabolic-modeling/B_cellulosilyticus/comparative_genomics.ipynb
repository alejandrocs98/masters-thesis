```python
import pandas as pd
```

```python
orthogroups = pd.read_table('Orthology/Orthogroups/Orthogroups.tsv', index_col=0, sep='\t')
orthogroups = orthogroups.fillna('')
orthogroups['B_cellulosilyticus_DSM_14838'] = orthogroups['B_cellulosilyticus_DSM_14838'].apply(lambda x: x.split(', '))
orthogroups['B_cellulosilyticus_WH2'] = orthogroups['B_cellulosilyticus_WH2'].apply(lambda x: x.split(', '))
orthogroups
```

```python
orthocount = pd.read_table('Orthology/Orthogroups/Orthogroups.GeneCount.tsv', index_col=0, sep='\t')
orthocount.drop('Total', axis=1, inplace=True)
orthocount
```

```python
DMS_annot = pd.read_table('Bacteroides_cellulosilyticus_DSM_14838/Bacteroides_cellulosilyticus_DSM_14838_annotations.tsv', index_col=0, sep='\t')
DMS_annot
```

```python
WH2_annot = pd.read_table('Bacteroides_cellulosilyticus_WH2/Bacteroides_cellulosilyticus_WH2_annotations.tsv', index_col=0, sep='\t')
WH2_annot
```

```python
wh2_up = orthocount[orthocount['B_cellulosilyticus_DSM_14838'] < orthocount['B_cellulosilyticus_WH2']].copy()
wh2_up['ratio'] = wh2_up['B_cellulosilyticus_WH2'].astype(str) + ':' + wh2_up['B_cellulosilyticus_DSM_14838'].astype(str)
wh2_up.drop(['B_cellulosilyticus_DSM_14838', 'B_cellulosilyticus_WH2'], axis=1, inplace=True)
wh2_up = pd.merge(wh2_up, orthogroups, left_index=True, right_index=True)
wh2_up.drop('B_cellulosilyticus_DSM_14838', axis=1, inplace=True)
wh2_up.rename(columns={'B_cellulosilyticus_WH2': 'genes'}, inplace=True)
wh2_up = wh2_up.explode('genes')
wh2_up.reset_index(inplace=True)
wh2_up.set_index('genes', inplace=True)
wh2_up
```

```python
wh2_down = orthocount[orthocount['B_cellulosilyticus_DSM_14838'] > orthocount['B_cellulosilyticus_WH2']].copy()
wh2_down['ratio'] = wh2_down['B_cellulosilyticus_DSM_14838'].astype(str) + ':' + wh2_down['B_cellulosilyticus_WH2'].astype(str)
wh2_down.drop(['B_cellulosilyticus_DSM_14838', 'B_cellulosilyticus_WH2'], axis=1, inplace=True)
wh2_down = pd.merge(wh2_down, orthogroups, left_index=True, right_index=True)
wh2_down.drop('B_cellulosilyticus_WH2', axis=1, inplace=True)
wh2_down.rename(columns={'B_cellulosilyticus_DSM_14838': 'genes'}, inplace=True)
wh2_down = wh2_down.explode('genes')
wh2_down.reset_index(inplace=True)
wh2_down.set_index('genes', inplace=True)
wh2_down
```

```python
wh2_down_annot = pd.merge(wh2_down, DMS_annot, left_index=True, right_index=True)
wh2_down_annot = wh2_down_annot[wh2_down_annot['EC'] != '-'].loc[:,['Orthogroup', 'ratio', 'EC', 'Description']]
wh2_down_annot = wh2_down_annot.drop_duplicates(subset=['Orthogroup', 'EC'])
wh2_down_annot
```

```python
wh2_up_annot = pd.merge(wh2_up, WH2_annot, left_index=True, right_index=True)
wh2_up_annot = wh2_up_annot[wh2_up_annot['EC'] != '-'].loc[:,['Orthogroup', 'ratio', 'EC', 'Description']]
wh2_up_annot = wh2_up_annot.drop_duplicates(subset=['Orthogroup', 'EC'])
wh2_up_annot
```

```python
with pd.ExcelWriter('differential_reactions.xlsx') as writer:
    # Save each DataFrame to a separate sheet
    wh2_up_annot.to_excel(writer, sheet_name='WH2_up', index=True)
    wh2_down_annot.to_excel(writer, sheet_name='WH2_down', index=True)
```
