```python
from cobra.io import read_sbml_model
from cobra import Model, Reaction, Metabolite
import cobra
```

# Getting started

```python
b_cel_dsm = read_sbml_model("Bacteroides_cellulosilyticus_DSM_14838.xml")
b_cel_dsm
```

```python
# Reactions
testr = b_cel_dsm.reactions.get_by_id('EX_MGlcn196_rl(e)')
testr
```

```python
print(testr.name)
print(testr.reaction)
print(testr.reactants)
print(testr.reversibility)
print(testr.bounds)
print(testr.check_mass_balance())
testr.add_metabolites({b_cel_dsm.metabolites.get_by_id("10fthf[c]"): -1})
print(testr.reaction)
testr.subtract_metabolites({b_cel_dsm.metabolites.get_by_id("10fthf[c]"): -1})
print(testr.reaction)
# test.reaction = 'h2o[c] + lcts[c] -> gal[c] + glc_D[c]'   to add a reaction
```

```python
# Metabolites
testm = b_cel_dsm.metabolites.get_by_id('glc_D[c]')
testm
```

```python
print(testm.name)
print(testm.compartment)
print(testm.charge)
print(testm.formula)
print(testm.reactions)
```
